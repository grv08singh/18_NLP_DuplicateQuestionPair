```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
import torch
from transformers import AutoTokenizer, AutoModel
from torch.utils.data import DataLoader
```

```python
df = pd.read_csv('data/02_preprocessed_data.csv')
df = df.dropna(subset=['question1', 'question2']).reset_index(drop=True)
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2', device='cuda')
```

```python
emb_q1 = model.encode(
    df['question1'].tolist(),
    batch_size=256,
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
emb_q2 = model.encode(
    df['question2'].tolist(),
    batch_size=256,
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
np.save('data/q1_emb_sbert.npy', emb_q1)
np.save('data/q2_emb_sbert.npy', emb_q2)
```
